```python
# Exercício 20 - O programa cria uma matriz original e depois inverte as colunas dela em linhas.
# Criei a matriz com 2 linhas e 3 colunas e inicializei a lista 'transposta' vazia para receber o resultado.
# O primeiro 'for' vai rodar o número de vezes equivalente às colunas da primeira linha usando o len(matriz[0]).
# Dentro dele, criei a lista 'nova_linha' que vai juntar temporariamente os números de cada coluna.
# O segundo 'for' serve para descer verticalmente pelas linhas da matriz original usando o len(matriz).
# Usei a função append() para pegar o número exato da linha e coluna e salvar na 'nova_linha', e depois usei append() de novo para mandar essa linha para'transposta'.
# No final, utilizei laços 'for' simples apenas para imprimir as duas matrizes organizadas na tela.

matriz = [[1, 2, 3], [4, 5, 6]]
transposta = []

for coluna in range(len(matriz[0])):
    nova_linha = []
    for linha in range(len(matriz)):
        nova_linha.append(matriz[linha][coluna])
    transposta.append(nova_linha)

print("Matriz original:")
for linha in matriz:
    print(linha)

print("Matriz transposta:")
for linha in transposta:
    print(linha)


```

```python
# Exercício 21 - Criei as variáveis 'inicio' e 'fim' pra marcar onde começa e onde termina a parte da lista que eu ainda vou olhar.
# O 'while' fica repetindo enquanto o 'inicio' não passar do 'fim', ou seja, enquanto ainda tiver lista pra procurar.
# Dentro dele eu pego o 'meio' da lista pra comparar com o número que a pessoa digitou.
# Se o número for igual ao valor do 'meio', ele foi encontrado e o programa encerra a busca.
# Se o número for maior, coloco o 'ínicio' para a metade da frente da lista.
# Se for menor, eu coloco o 'fim' para a metade de trás da lista.
# Usei a variável 'encontrado' só pra guardar se achei ou não.

numeros = [2, 5, 8, 10, 15, 20, 25]
procurado = int(input("Digite o número que deseja encontrar: "))

inicio = 0
fim = len(numeros) - 1

encontrado = False

while inicio <= fim:
  meio = (inicio + fim) // 2
  if numeros[meio] == procurado:
    encontrado = True
    break
  elif procurado > numeros[meio]:
   inicio = meio + 1
  else:
    fim = meio - 1
  if encontrado:
    print("Número encontrado na lista.")
  else:
     print("Número não encontrado.")
```

```python
# Exercício 22 - O programa pede para o usuário digitar uma mensagem.
# Criei a variável 'deslocamento' com o valor 5, pois cada letra será deslocada cinco posições.
# Depois criei a variável 'resultado' para guardar a nova mensagem criptografada.
# O 'for' percorre a mensagem letra por letra.
# O 'isalpha()' é reponsável por verificar se o caractere é uma letra. Se for, o programa continua a criptografia.
# O 'ord()' é o respnsável por transformar a letra em um número da tabela Unicode.
# Em seguida, o programa soma o valor do deslocamento para avançar cinco posições.
# O 'chr()' transforma esse número novamente em uma letra.
# Se o caractere não for uma letra, como espaço ou pontuação, ele permanece igual.

mensagem = input("Digite uma mensagem: ")
deslocamento = 5
resultado = ""

for letra in mensagem:
  if letra.isalpha():
      nova = chr(ord(letra) + deslocamento)
      resultado += nova
  else:
      resultado += letra
print("Mensagem criptografada:", resultado)

```

```python
# Exercício 23 - # Criei a variável 'deslocamento' com o valor 5, pois será utilizada para voltar as letras em cinco posições.
# Depois criei a variável 'resultado' para guardar a mensagem descriptografada.
# O 'for' percorre a mensagem letra por letra.
# O 'isalpha()' é responsável por verificar se o caractere é uma letra. Se for, o programa continua o processo.
# O 'ord()' é o responsável por transformar a letra em um número da tabela Unicode.
# Em seguida, o programa subtrai o valor do deslocamento para retornar à letra original.
# O 'chr()' transforma esse número novamente em uma letra.
# Se o caractere não for uma letra, como espaço ou pontuação, ele permanece igual.
# O exercício é praticamente igual ao 22, o que muda que subtrai ao invés de somar o deslocamento

mensagem = input("Digite uma mensagem: ")
deslocamento = 5
resultado = ""

for letra in mensagem:
  if letra.isalpha():
      nova = chr(ord(letra) - deslocamento)
      resultado += nova
  else:
      resultado += letra
print("Mensagem original:",resultado)
```

```python
# Exercício 31 - Primeiro criei uma lista com alguns números.
# Depois, utilizei dois laços 'for' para percorrer e comparar os números da lista.
# O primeiro 'for' serve para percorrer toda lista e o segundo serve para enquanto vai percorrendo a lista, vai comparando ao mesmo tempo.
# Se o número da esquerda for maior que o da direita, os dois trocam de posição.
# O processo se repete até todos os números das listas ficarem em ordem crescente.

numeros = [23, 1, 7, 99, 34]

for i in range(len(numeros)):
    for a in range(len(numeros) - 1):
        if numeros[a] > numeros[a + 1]:
            numeros[a], numeros[a + 1] = numeros[a + 1], numeros[a]

print("Lista ordenada:", numeros)
```

```python
# Exercício 36 - Primeiro importei as bibliotecas 'random' e 'string'.
# A biblioteca 'string' fornece letras e números que podem ser usados na senha., Usei ela para não preciar digitar muitos caracteres.
# Usei 'string.ascii_letters' pois ela possui todas as letras do alfabeto e 'string.digits' possui todos os números.
# Depois criei uma variável 'senha' vazia para armazenar os caracteres escolhidos.
# O 'for' repete a escolha dos caracteres 7 vezes para formar a senha.

import random
import string

caracteres = string.ascii_letters + string.digits

senha = ""

for _ in range(7):
    senha += random.choice(caracteres)

print("A senha gerada é :", senha)
```

```python
# Exercício 37 -  Primeiro criei uma função que chama ela mesma durante a execução, para calcular o fatorial de um número.
# O 'if' verifica se o número é 0 ou 1, pois nesses casos o resultado do fatorial é 1.
# Se não for 0 nem 1, o número é multiplicado pelo resultado do fatorial do número anterior.

def fatorial(n):
    if n == 0 or n == 1:
        return 1
    return n * fatorial(n - 1)

num = int(input("Digite um número: "))
print(f"O fatorial de {num} é {fatorial(num)}")
```

```python
# Exercício 40 - Primeiro, criei uma lista vazia de compras, o usuário poderá adicionar qualquer item.
# Depois, utilizei um laço de repetição 'while True', que apresenta as opções de adicionar, remover e listar os itens.
# Na opção 1, o programa pede para o usuário adicionar um item na lista. Utilizei o 'append()', pois ele adiciona o item à lista.
# Na opção 2, o usuário pode remover um item. Se o item estiver na lista, ele será removido pelo método 'remove'
# Caso o item não seja encontrado na lista, aparece uma mensagem dizendo que o item não foi encontrado
# Na opção 3, o programa mostra a lista atual com os itens que o usuário adicionou.

compras = []

while True:
    opcao = input("1-Adicionar - 2-Remover - 3-Listar: ")

    if opcao == "1":
        compras.append(input("Digite um item para adicionar à lista de compras: "))
    elif opcao == "2":
        item = input("Digite um item para remover da lista de compras: ")
        if item in compras:
            compras.remove(item)
        else:
            print(" O item não foi encontrado. Digite o item certo.")
    elif opcao == "3":
        print("Sua lista atual:", compras)

```

```python
# Exercício 41 - Primeiro criei uma lista com alguns números repetidos.
# Depois, eu criei uma lista vazia 'remover_duplicatas' para que quando os números repetidos forem removidos, apareçam na lista
# O laço 'for in' é responsável por percorrer os números da lista.
# O 'if' verifica se o número ainda não está na nova lista. Se não estiver, ele é adicionado.

numeros = [1, 1, 2, 3, 2, 9, 4, 9, 4, 1, 5, 5,  3]
remover_duplicatas = []

for numero in numeros:
    if numero not in remover_duplicatas:
        remover_duplicatas.append(numero)

print(" lista sem duplicatas é: ", remover_duplicatas)
```

```python
# Exercício 42 - O programa solicta que o usuário digite um número.
# O primeiro 'for in' é responsável por verificar a quantidade de números primos até o número que o usuário digitou, começando no número 2.
# O segundo 'for in' é responsável por verificar se o número possui mais de 1 divisor.
# Se encontrar um divisor, 'primo' recebe False e a verificação para.
# Se não encontrar nenhum divisor, o número é primo.

num = int(input("Digite um número: "))

for n in range(2, num + 1):
    primo = True
    for i in range(2, n):
        if n % i == 0:
            primo = False
            break
    if primo:
        print(n, end=" ")
```

```python
# Exercício 44 - O programa pede pro usuário informar quantos termos da sequência deseja ver.
# Depois, criei as duas primeiras variáveis da sequência, começando com 0 e 1.
# Utilizei um laço 'for' para repetir o programa a quantidade de vezes solicitada.
# Depois atualizei os valores, fazendo o próximo número ser a soma dos dois anteriores.

n = int(input("Digite o número da sequencia Fibonacci você deseja ver? "))

variavel_1 = 0
variavel_2 = 1

for _ in range(n):
    print(variavel_1, end=" ")
    variavel_1, variavel_2 = variavel_2, variavel_1 + variavel_2
```

```python
# Exercício 45 - Criei uma lista com alguns números pares e ímpares.
# O laço 'for' verifica os números presentes na lista.
# O 'if' serve para verificar se o numero dividido por 2 é 0. Se sim, é um número par.
# Se não, ele é um número ímpar.

numeros = [10, 7, 23, 12, 8, 5, 38]

for n in numeros:
    if n % 2 == 0:
        print("O número é par:", n)
    else:
        print("O número é ímpar:", n)
```

```python
# Exercício 46 - Criei uma lista com alguns números.
# Depois, utilizei a biblioteca 'statistics' para calcular a média e o desvio padrão.
# A função 'mean()' calcula a média e a função 'stdev()' calcula o desvio padrão.

import statistics

numeros = [6, 7, 25, 12, 8, 5, 38]

media = statistics.mean(numeros)
desvio_padrao = statistics.stdev(numeros)

print("A média é: ", media)
print("O desvio padrão é: ", desvio_padrao)
```

```python
# Exercício 47 - Criei uma lista com alguns números repetidos.
# Criei um dicionário vazio, que serve para guardar os números.
# O 'for' percorre todos os números da lista.
# O 'if' verifica se o número já está no dicionário. Se estiver, é somado mais 1 do mesmo.
# Se não, é adicionado o número pela primeira vez.


numeros = [1, 1, 1, 2, 3, 1, 4, 6, 3, 5, 4, 6]

contagem = {}

for numero in numeros:
    if numero in contagem:
        contagem[numero] += 1
    else:
        contagem[numero] = 1

print("Quantidade de cada número:", contagem)
```

```python
# Exercício 48 - Criei uma lista com números fora de ordem.
# O 'for' pega cada número da lista a partir do segundo.
# O programa compara o número com o que está antes dele.
# Se o número anterior for maior, o programa faz a troca de posição.
# O 'while' é o responsável por fazer as trocas de posição até que fique na ordem correta.

numeros = [3, 2, 8, 7, 9, 1, 5]

for posicao in range(1, len(numeros)):
    while posicao > 0 and numeros[posicao] < numeros[posicao - 1]:
        numeros[posicao], numeros[posicao - 1] = numeros[posicao - 1], numeros[posicao]
        posicao = posicao - 1

print("A lista ordenada é:", numeros)
```

```python
# Exercício 49 - O programa pede para o usuário digite um número.
# Criei a váriavel 'soma' para guardar a soma dos digitos.
# O laço 'for' percorre os números digitados.
# O 'int' é responsável por transformar cada digito em número.
# O programa soma os digitos na váriavel 'soma'.

numero = input("Digite um número: ")
soma = 0

for soma_digito in numero:
    soma += int(soma_digito)

print("A soma dos dígitos é: ", soma)
```

```python
# Exercício 50 - O programa pede para que o suário digite uma palavra.
# Usei a função '.lower' para colocar as letras da palavra eem forma minúscula.
# O laço 'for' é responsável por percorrer cada letra da palavra.
# O 'count()' é responsável por contar quantas vezes cada letra aparece na palavra.

palavra = input("Digite uma palavra: ").lower()

for letra in palavra:
    if letra.isalpha():
        print(letra, ":", palavra.count(letra))
```
