```python
import numpy as np
from scipy import integrate, linalg, optimize, interpolate
from scipy.fft import fft, ifft

np.random.seed(42)
```

```python
# Exercício 1 -
f1 = lambda x: x**2 * np.sin(x)
resultado, erro = integrate.quad(f1, 0, np.pi)
print("Integral:", resultado, "| erro estimado:", erro)
```

```python
# Exercício 2 -
edo = lambda x, y: x - y
sol = integrate.solve_ivp(edo, [0, 10], [1], t_eval=np.linspace(0, 10, 11))
print("x:", np.round(sol.t, 2))
print("y:", np.round(sol.y[0], 4))
```

```python
# Exercício 3 -
raizes = np.roots([1, -3, 0, 0, 2])
print("Raízes:", raizes)
```

```python
# Exercício 4 -
neg_f = lambda x: -(-x**2 + 4*x + 1)
res = optimize.minimize(neg_f, 0)
print("x que maximiza:", res.x[0], "| valor máximo:", -res.fun)
```

```python
# Exercício 5 -
t = np.linspace(0, 1, 500, endpoint=False)
sinal = np.sin(2*np.pi*t) + np.cos(4*np.pi*t)
transformada = fft(sinal)
frequencias = np.fft.fftfreq(len(t), d=t[1]-t[0])
mag = np.abs(transformada)
picos = np.argsort(mag)[-4:]
print("Frequências dominantes (Hz):", np.round(frequencias[picos], 1))
```

```python
# Exercício 6 -
A = np.random.rand(5, 5)
inversa = linalg.inv(A)
print("Verificação (A·A⁻¹ ≈ identidade):\n", np.round(A @ inversa, 6))
```

```python
# Exercício 7 -
x7 = np.array([0, 1, 2, 3, 4])
y7 = np.array([0, 2, 3, 5, 4])
f_interp = interpolate.interp1d(x7, y7, kind='cubic')
novos = np.array([0.5, 1.5, 2.5, 3.5])
print("Valores interpolados em", novos, ":", np.round(f_interp(novos), 4))
```

```python
# Exercício 8 -
A8 = np.array([[3, 2], [1, 4]])
b8 = np.array([5, 6])
sol8 = linalg.solve(A8, b8)
print("Solução [x, y]:", sol8)
```

```python
# Exercício 9 -
x9 = np.array([1, 2, 3, 4, 5])
y9 = np.array([2, 4, 5, 7, 8])
parabola = lambda x, a, b, c: a*x**2 + b*x + c
params, _ = optimize.curve_fit(parabola, x9, y9)
print("Parâmetros [a, b, c]:", np.round(params, 4))
```

```python
# Exercício 10 -
res10, err10 = integrate.dblquad(lambda y, x: x*y + x**2, 0, 2, 0, 1)
print("Resultado:", res10, "| erro:", err10)
```

```python
# Exercício 11 -
M = np.random.randint(1, 10, size=(4, 4)).astype(float)
autovalores, autovetores = linalg.eig(M)

print("Matriz:", M.astype(int))
print("Autovalores:", np.round(autovalores, 4))
print("Autovetores (colunas):", np.round(autovetores, 4))
```

```python
# Exercício 12 -
g = lambda x: np.exp(x) - x**2
raiz = optimize.fsolve(g, 1)
print("Raiz encontrada:", raiz)
```

```python
# Exercício 13 -
M13 = np.random.uniform(-10, 10, size=(6, 6))
print("Determinante:", linalg.det(M13))
```

```python
# Exercício 14 -
f14 = lambda v: (v[0]-2)**2 + (v[1]-3)**2
res14 = optimize.minimize(f14, [0, 0])
print("Mínimo em [x, y]:", np.round(res14.x, 4))
```

```python
# Exercício 15 -
sinal_complexo = np.random.rand(8) + 1j*np.random.rand(8)
inversa_fft = ifft(sinal_complexo)

print("Sinal (frequência):", np.round(sinal_complexo, 3))
print("IFFT (tempo):", np.round(inversa_fft, 4))
```

```python
# Exercício 16 -
def sistema(x, estado):
    u, v = estado
    return [v, -u]
sol16 = integrate.solve_ivp(sistema, [0, 10], [1, 0], t_eval=np.linspace(0, 10, 6))

print("x:", np.round(sol16.t, 2))
print("y (numérico):", np.round(sol16.y[0], 4))
print("cos(x) (esperado):", np.round(np.cos(sol16.t), 4))
```

```python
# Exercício 17 -
raiz_cubica = optimize.newton(lambda x: x**3 - 27, 2)
print("Raiz cúbica de 27:", raiz_cubica)
```

```python
# Exercício 18 -
x18 = np.array([1, 2, 3, 4, 5])
y18 = x18**2
spline = interpolate.CubicSpline(x18, y18)
teste = np.array([1.5, 2.5, 3.5, 4.5])

print("Spline em", teste, ":", np.round(spline(teste), 4))
print("x² real:", teste**2)
```

```python
# Exercício 19 -
n = 64
t19 = np.linspace(0, 1, n, endpoint=False)
sinal19 = np.sin(2*np.pi*5*t19) + np.cos(2*np.pi*10*t19)
fft19 = fft(sinal19)
freqs19 = np.fft.fftfreq(n, d=t19[1]-t19[0])
picos19 = np.argsort(np.abs(fft19))[-4:]
print("Frequências dominantes (Hz):", np.sort(np.abs(np.round(freqs19[picos19], 1))))
```
